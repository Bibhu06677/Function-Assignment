**1. What is the difference between a function and a method in Python?**

In Python, functions and methods are both blocks of code that perform a specific task and can be called (invoked) by name, but they differ mainly in how they are associated with objects.

a. Function
A function is a block of code that is defined using def and is not associated with any object (although it can still work with objects as parameters). It can be called independently.

Example of a Function:

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Bibhu"))
```

b. Method
A method is a function that is associated with an object — it is defined inside a class and is called on an instance of that class. The first argument of a method is usually self, which refers to the instance calling the method.

Example of a Method:

```python
class Greeter:
    def greet(self, name):
        return f"Hello, {name}!"

greeter = Greeter()
print(greeter.greet("Bibhu"))
```

Key Differences:

| Aspect            | Function                                 | Method                                        |
|-------------------|------------------------------------------|-----------------------------------------------|
| Defined with      | `def` outside any class                  | `def` inside a class                          |
| Called on         | Its own, without a class/instance        | An object (instance of a class)               |
| First parameter   | Normal parameters                        | Typically `self` (the instance itself)        |
| Example call      | `greet("Bibhu")`                         | `greeter.greet("Bibhu")`                        |

**2. Explain the concept of function arguments and parameters in Python.**

In Python, parameters and arguments refer to values used in functions, but they are used in different contexts:

Parameters
Parameters are the placeholders listed in the function definition. They define what kind of data the function can receive.

Arguments
Arguments are the actual values passed to a function when it is called.

Example:

```python
# Function definition with parameters
def greet(name, message):
    print(f"Hello {name}, {message}")

# Function call with arguments
greet("Arjun", "good morning!")
```

Explanation:
name and message are parameters (used in the definition).

"Arjun" and "good morning!" are arguments (used in the function call).

**5. What are iterators in Python and how do they differ from iterables?**

These two concepts are closely related but not the same.

Iterable
An iterable is any object that can be looped over (e.g., in a for loop).
Examples: lists, tuples, strings, dictionaries, sets.

It implements the __iter__() method.

```python
my_list = [1, 2, 3]

for item in my_list:  # my_list is an iterable
    print(item)
```

Iterator
An iterator is an object that remembers its position and returns items one at a time using the next() function.

It implements both:

__iter__() → returns itself

__next__() → returns the next item or raises StopIteration

You can get an iterator from an iterable using the iter() function.

```python
my_list = [1, 2, 3]
it = iter(my_list)  # Create an iterator from a list

print(next(it))  # 1
print(next(it))  # 2
print(next(it))  # 3
# print(next(it))  # Raises StopIteration
```

Key Differences:

| Feature      | Iterable           | Iterator                |
|--------------|--------------------|--------------------------|
| What it is   | Can be looped over | Gives items one at a time |
| Methods      | `__iter__()`       | `__iter__()` and `__next__()` |
| Examples     | List, tuple, string | Output of `iter()`     |
| Reusability  | Can be reused       | Used once unless reset  |

**6. Explain the concept of generators in Python and how they are defined.**

A generator is a special type of function that yields values one at a time instead of returning them all at once. This makes them memory-efficient, especially for large data sets or infinite sequences.

How Are Generators Defined?
Generators are defined like normal functions but use the yield keyword instead of return.

Example: A Simple Generator

```python
def count_up_to(max):
    count = 1
    while count <= max:
        yield count
        count += 1

# Using the generator
counter = count_up_to(3)
print(next(counter))  # 1
print(next(counter))  # 2
print(next(counter))  # 3
# print(next(counter))  # Raises StopIteration
```

Explanation:
count_up_to is a generator function.

Each time yield is hit, the function pauses and remembers where it left off.

The next call to next() resumes from that point.

**7. What are the advantages of using generators over regular functions?**

Generators offer several key benefits, especially when working with large data or streams:

1. Memory Efficiency
Generators yield one item at a time instead of creating and storing the entire result in memory.

 Example:

```python
def count_up_to(n):
    for i in range(n):
        yield i

# Uses very little memory
for num in count_up_to(1_000_000):
    pass
```

2. Faster Startup Time
Generators start producing values immediately, while regular functions wait until all results are prepared.

3. Can Model Infinite Sequences

Generators can produce infinite data streams without running out of memory.

Example:

```python
def infinite_numbers():
    num = 0
    while True:
        yield num
        num += 1
```

4. Cleaner and More Readable Code
Generators simplify code when dealing with data pipelines or complex iteration logic.

5. State Preservation
Each yield pauses the function and saves its state, so it can resume where it left off.

**8. What is a lambda function in Python and when is it typically used?**

A lambda function is a small anonymous function defined with the lambda keyword. It can take any number of arguments but has only one expression.

Syntax:

lambda arguments: expression


```python
square = lambda x: x * x
print(square(5))
```

When Is a Lambda Function Typically Used?

Lambda functions are mostly used:

For short, simple operations

When a function is needed temporarily, especially as an argument to higher-order functions

Common Use Cases:

1. With map()

```python
nums = [1, 2, 3, 4]
squares = list(map(lambda x: x * x, nums))
print(squares)
```

2. With filter()

```python
nums = [1, 2, 3, 4, 5]
even = list(filter(lambda x: x % 2 == 0, nums))
print(even)
```

3. With sorted()

```python
names = ["Alice", "Bob", "Charlie"]
sorted_names = sorted(names, key=lambda x: len(x))
print(sorted_names)
```

**9. Explain the purpose and usage of the `map()` function in Python.**

map() Function in Python

The map() function is used to apply a function to every item in an iterable (like a list or tuple) and return a new map object (which is an iterator).

Syntax:

map(function, iterable)

function: A function to apply (can be regular or lambda)

iterable: A sequence (like list, tuple, etc.)

Example: Using map() with a Regular Function



```python
def square(x):
    return x * x

numbers = [1, 2, 3, 4]
result = map(square, numbers)
print(list(result))
```

Example: Using map() with Lambda Function

```python
numbers = [1, 2, 3, 4]
squares = list(map(lambda x: x ** 2, numbers))
print(squares)
```

Example: Mapping Multiple Iterables

```python
a = [1, 2, 3]
b = [4, 5, 6]
result = list(map(lambda x, y: x + y, a, b))
print(result)
```

Purpose and Benefits:

Simplifies applying a function to each item in a collection

More concise and often faster than a for loop

Works well with lambda for inline transformations

**10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?**

These three functions are all used to process iterables (like lists or tuples) in Python, but they do so in different ways. Let's go through each one:

1. map() Function

The map() function applies a function to every item in an iterable and returns a new iterator with the results.

Purpose: Modify each element of an iterable.

Usage: Ideal for transformations where you want to modify every item.

Syntax:

map(function, iterable)

Example:

```python
numbers = [1, 2, 3, 4]
result = list(map(lambda x: x * 2, numbers))
print(result)
```

2. reduce() Function (from functools module)

The reduce() function applies a function cumulatively to the items in an iterable, so as to reduce it to a single value. It takes two arguments at a time and applies the function recursively.

Purpose: Accumulate or reduce the iterable into a single value.

Usage: Useful for operations like summing values, finding products, etc.

Syntax:

from functools import reduce
reduce(function, iterable)

Example:

```python
from functools import reduce

numbers = [1, 2, 3, 4]
result = reduce(lambda x, y: x + y, numbers)
print(result)
```

3. filter() Function

The filter() function filters the elements of an iterable based on a boolean condition defined by a function. It returns an iterator with elements that satisfy the condition.

Purpose: Filter out elements based on a condition (true/false).

Usage: Ideal for keeping only the items that match a condition.

Syntax:

filter(function, iterable)

Example:

```python
numbers = [1, 2, 3, 4, 5]
result = list(filter(lambda x: x % 2 == 0, numbers))
print(result)
```

 Comparison Table:

 | Function    | Purpose                                      | Returns             | Example Usage                          |
|-------------|----------------------------------------------|---------------------|----------------------------------------|
| `map()`     | Applies a function to all items              | Transformed iterable | Square all numbers in a list          |
| `reduce()`  | Cumulatively applies a function to the iterable | A single value      | Sum of all numbers in a list          |
| `filter()`  | Filters the iterable based on a condition    | Filtered iterable   | Get only even numbers from a list      |

5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

```python
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.a
fib = FibonacciIterator(10)
for num in fib:
    print(num, end=' ')

```

6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

```python
def powers_of_two(max_exponent):
    for exp in range(max_exponent + 1):
        yield 2 ** exp
for power in powers_of_two(5):
    print(power)
```

9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

```python
# Celsius to Fahrenheit conversion formula: (°C × 9/5) + 32
def c_to_f(celsius):
    return (celsius * 9/5) + 32

# Sample list of temperatures in Celsius
celsius_temps = [0, 20, 37, 100]

# Convert using map()
fahrenheit_temps = list(map(c_to_f, celsius_temps))

print(fahrenheit_temps)
```
